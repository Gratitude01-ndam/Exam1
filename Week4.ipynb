# Automated Quality Control Application

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Load base model
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))

# Build model
model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dense(2, activation='softmax') # Assuming 2 classes: Healthy and Defective
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Model compiled successfully. You can train this model using an appropriate dataset.")
print("For this demonstration, we'll proceed as if the model is already trained and deploy it using Streamlit.")
```

### Create Streamlit Application File

Now, let's create the `app.py` file with your Streamlit code. This file will be used to deploy a simple web interface for your model.

```python
%%writefile app.py
import streamlit as st
from PIL import Image
import numpy as np
from tensorflow.keras.models import load_model # Assuming model will be loaded or passed globally
import tensorflow as tf

# To make the model accessible to Streamlit, we need to either load it or define it within the app.py
# For simplicity, let's redefine a dummy model structure or assume it's loaded from a saved file.
# In a real scenario, you would save your trained model and load it here.

# For demonstration purposes, we'll redefine the model structure here.
# If you have a trained model, save it and load using model = load_model('your_model.h5')
# To ensure the model definition matches the one above:
def get_model():
    base_model = tf.keras.applications.MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
    model = tf.keras.models.Sequential([
        base_model,
        tf.keras.layers.GlobalAveragePooling2D(),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(2, activation='softmax')
    ])
    # In a real application, you would load pre-trained weights here
    # For this example, we'll use an untrained model for prediction structure
    return model

model = get_model()

st.title("Automated Quality Control")
uploaded_file = st.file_uploader("Upload an image", type=["jpg","png","jpeg"])

if uploaded_file:
    image = Image.open(uploaded_file).resize((224,224))
    st.image(image, caption="Uploaded Image", use_column_width=True)
    img_array = np.array(image)/255.0
    img_array = np.expand_dims(img_array, axis=0)

    # Make prediction
    # Since the model is untrained here, predictions will be random.
    # In a real scenario, `model.predict` would yield meaningful results.
    pred = model.predict(img_array)
    label = "Healthy" if np.argmax(pred) == 0 else "Defective"
    st.write(f"**Prediction:** {label}")

# Add a reload button
if st.button("Reload App"):
    st.experimental_rerun()
```
